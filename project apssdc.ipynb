### Queries

#### 1)Which state had the highest total over all year?

```python
max_total_row = data.loc[data['Total'].idxmax()]
print("State with highest total:",max_total_row['STATE/UT'])
print("Total_value:",max_total_row['Total'])
```

#### 2)Which state had the lowest total?

```python
min_total_row = data.loc[data['Total'].idxmin()]
print("State with highest total:",min_total_row['STATE/UT'])
print("Total_value:",min_total_row['Total'])
```

#### 3)What is the average 'Total' across all states?

```python
avg_total = data['Total'].mean()
print("Average total across states: ",avg_total)
```

#### 4)Which year had the highest combined value across all states?

```python
year_columns = [col for col in data.columns if col.isdigit()]
year_sums = data[year_columns].sum()
max_year = year_sums.idxmax()
print(f"Year with highest value overall: {max_year},value:{year_sums[max_year]}")
```

#### 5)There are inordered years are left over how could you arrange them?

```python
year_cols = [col for col in data.columns if str(col).isdigit()]
sorted_year = sorted(year_cols,key=int)
imp_cols = ['Total','Growth','Peak_year','category']
non_year_cols = [col for col in data.columns if col not in year_cols+imp_cols]
data = data[non_year_cols+sorted_year+imp_cols]
data.columns

```

#### 6)Which state had the biggest growth between any two years?

```python
data['Growth']=data[year_cols].max(axis=1)-data[year_cols].min(axis=1)
max_growth=data.loc[data['Growth'].idxmax()]
print("state with highest growth: ",max_growth['STATE/UT'])
print("Growth: ",max_growth['Growth'])
```

#### 7)Top 5 states by peak value and their peak years?

```python
data['Peak_val']=data[year_cols].max(axis=1)
data['Peak_year']=data[year_cols].idxmax(axis=1)
top5 = data.sort_values(by='Peak_val',ascending = False).head(5)
print(top5[['STATE/UT','Peak_val','Peak_year']])
```

#### 8)Which state had the sharpest fall between two consecutive years?

```python
diff = data[year_cols].diff(axis=1)
min_fall=data.loc[diff.min(axis=1).idxmin()]
print("State with sharpest fall: ",min_fall['STATE/UT'])
```

#### 9)which state had zero values for most years?

```python
zero_case = (data[year_cols] == 0).sum(axis=1)
max_zero = data.loc[zero_case.idxmax()]
print("State with most zeroes: ",max_zero['STATE/UT'],"--",zero_case.max(),'years')
```

#### 10)How many states had their peak in 2020?

```python
peak_of_2020=data[data['Peak_year']=='2020']
print("State with peak in 2020: ",len(peak_of_2020))
print(peak_of_2020['STATE/UT'])
```

#### 11)Rank the states by median value over all years? 

```python
data['Median'] = data[year_cols].median(axis=1)
rank = data[['STATE/UT','Median']].sort_values(by = 'Median',ascending = False)
print(rank)
```

#### Advance Stats

```python
year_cols = [col for col in data.columns if col.isdigit()]
year_cols
```

```python
data_long = data.melt(id_vars=['STATE/UT'],value_vars=year_cols,var_name='Year',value_name='Cases')
data_long["Year"] = data_long["Year"].astype(int)   
data_long = data_long.sort_values(by=['STATE/UT','Year']).reset_index(drop=True)
data_long
```

```python
data_long.isna().sum()
```

```python
data_long["Cases"] = data_long["Cases"].fillna(0)
data_long.dtypes
```

```python
data_long["Cases"] = pd.to_numeric(data_long["Cases"],errors='coerce')
```

```python
national = data_long.groupby('Year')['Cases'].sum()
national
```
